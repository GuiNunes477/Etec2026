```python
#QUESTÃO 02 – Soma dos Números (2 pontos) solicite ao usuário um número inteiro positivo. Utilizando um laço de repetição, calcule e exiba a soma de todos os números de 1 até o número informado.

numero1 = int(input("Escreva um número inteiro positivo "))
numero = 1
i=1
for i in range(1 , numero1+1):
  fatorial = (numero + (i + 1))
  print(fatorial)
```

```python

# Um caixa eletrônico possui notas de: R$ 100 R$ 50 R$ 20 R$ 10 R$ 5 R$ 2 R$ 1 Solicite ao usuário o valor do saque e informe a menor quantidade possível de notas para realizar o pagamento.

cem = 0
cinq = 0
vinte = 0
dez = 0
cinco = 0
dois = 0
um = 0
dinheiro = int(input("Qual o valor do saque? "))
while dinheiro != 0:
  if dinheiro >= 100:
   dinheiro -= 100
   cem +=1
  elif dinheiro >= 50:
   dinheiro -= 50
   cinq += 1
  elif dinheiro >= 20:
   dinheiro -= 20
   vinte += 1
  elif dinheiro >=10:
   dinheiro -= 10
   dez += 1
  elif dinheiro >=5:
   dinheiro -= 5
   cinco += 1
  elif dinheiro >= 2:
   dinheiro -= 2
   dois += 1
  elif dinheiro >= 1:
   dinheiro -= 1
   um += 1
  print(f"Notas de 100 = {cem}")
  print(f"Notas de 50 = {cinq}")
  print(f"Notas de 20 = {vinte}")
  print(f"Notas de 10 = {dez}")
  print(f"Notas de 5 = {cinco}")
  print(f"Notas de 2 = {dois}")
  print(f"Notas de 2 = {um}")
```
