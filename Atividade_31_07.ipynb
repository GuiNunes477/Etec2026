```python
# Exercício 3 – Operações Matemáticas
numero = float(input("insira um número: "))
numero2 = float(input("insira outro número: "))
print(f" A soma de {numero} com {numero2} é {numero + numero2}, a subtração é {numero - numero2}, A multiplicação é {numero * numero2} e a divisão é {numero / numero2}")
```

```python
# Exercício 4 – Média de Notas
nota1 = float(input("Qual foi a sua primeira nota? "))
nota2 = float(input("Qual foi a sua segunda nota? "))
nota3 = float(input("Qual foi a sua terceira nota? "))
media = (nota1 + nota2 + nota3) / 3
if(media <= 10 and media >=7):
 print(f"Parabéns, sua media é {round(media,2)} você passou")
elif(media >=5 and media <=6,9):
 print(f"Sua média foi {round(media,2)} você está de recuperação")
elif(media <5 and media >= 0):
  print(f"sua média foi {round(media,2)} você reprovou")
elif(media < 0 or media > 10):
  print("nota invalida")
```

```python
# Exercício 7 – Número Par ou Ímpar
numero = float(input("Insira um número: "))
if(numero % 2 == 0):
  print(f"O número {numero} é par")
else:
  print(f"O número {numero} é ímpar")
```

```python
# Exercício 8 – Maior Número
numero = float(input("Insira um número: "))
numero2 = float(input("Insira um número: "))
numero3 = float(input("Insira um número: "))
if(numero > numero2 and numero > numero3):
  print(f"{numero} é maior que {numero2} e {numero3}")
elif(numero2 > numero and numero2 > numero3):
  print(f"{numero2} é maior que {numero} e {numero3}")
elif(numero3 > numero and numero3 > numero ):
  print(f"{numero3} é maior que {numero} e {numero2}")
elif(numero == numero2 and numero == numero3):
  print(f"Todos os números são iguais {numero}")
```

```python
# Exercício 9 – Tabuada
numero = int(input("Insira um número de 1 a 10: "))
for i in range(1, 11):
  if(numero <= 10 and numero >= 1):
    print(f"{numero} * {i} = {numero * i}")
  else:
    print(f"Número Invalido")
```

```python
# Exercício 10 – Calculadora Simples
numero = float(input("Insira um número: "))
numero2 = float(input("Insira outro número: "))
operacoes = input("Escolha uma das operaçõeas entre adição(+), subtração(-), Multiplicação(*) ou divisão(/) ")
if(operacoes == "+"):
  print(f"{numero} + {numero2} = {round(numero + numero2,2)}")
elif(operacoes == "-"):
  print(f"{numero} - {numero2} = {round(numero - numero2,2)}")
elif(operacoes == "*"):
  print(f"{numero} * {numero2} = {round(numero * numero2,2)}")
elif(operacoes == "/"):
  print(f"{numero} / {numero2} = {round(numero / numero2,2)}")

```
