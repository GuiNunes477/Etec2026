```python
# Exercício 1 – Idade para dirigir
idade = int(input("Quantos anos você tem? "))
if(idade >= 18):
  print("Você pode dirigir")
else:
  print("Você precisa ter 18 anos para dirigir")
```

```python
# Exercício 2 – Par ou Ímpar
numero = float(input("Insira um número: "))
if(numero % 2 == 0):
  print("Seu número é par")
else:
  print("Seu número é ímpar")
```

```python
# Exercício 3 – Maior de dois números
numero = float(input("Insira um número: "))
numero2 = float(input("Insira outro número: "))
if(numero > numero2):
  print(f"O número {numero} é maior que {numero2}")
else:
  print(f"O número {numero2} é maior que {numero}")
```

```python
# Exercício 4 – Situação do aluno
nota = int(input("Insira sua nota: "))
if(nota >= 7 and nota <= 10):
  print("Parabéns você passou")
elif(nota >= 5 and nota < 7):
  print("Você está de recuperação")
elif(nota < 0 or nota > 10):
  print("nota invalida")
else:
  print("Você está de reprovado")
```

```python
# Exercício 5 – Desconto na compra
valor = float(input("Qual o valor da compra? "))
if(valor >= 100):
  print(f"Você recebeu um desconto de 10%, o valor final é {round(valor - (valor * 0.10),2)}")
```

```python
# Exercício 6 – Positivo, Negativo ou Zero
numero = float(input("Insira um número: "))
if(numero > 0):
  print(f"O número {numero} é positivo")
elif(numero < 0):
  print(f"O número {numero} é negativo")
else:
  print(f"O número {numero} é zero")
```

```python
# Exercício 7 – Classificação de idade
idade = int(input("Quantos anos você tem? "))
if(idade < 12 and idade >= 0):
  print("Você é criança")
elif(idade >= 12 and idade <= 17):
  print(f"Você é adolescente")
elif(idade >= 18 and idade <= 59):
  print("Você é adulto")
elif(idade < 0):
  print("impossivel, você nem nasceu?")
else:
  print("Você é idoso")
```

```python
# Exercício 8 – Média de duas notas
nota = int(input("insira sua nota "))
nota2 = int(input("insira sua segunda nota "))
media = (nota + nota2)/2
if(media >= 7 and media <= 10):
  print(f"Parabéns você passou sua nota foi {round(media,2)}")
elif(media < 7 and media > 0):
  print(f"Você reprovou sua nota foi {round(media,2)}")
else:
  print("Nota invalida")
```

```python
# Exercício 9 – Login simples
usuario = input("insira o usuario: ")
senha = input("Insira sua senha: ")
if(senha == "1234" and usuario == "admin"):
  print("Bem vindo admin")
else:
  print("Senha ou usuario incorretos")
```

```python
# Exercício 10 – Calculadora de IMC
massa = float(input("Qual o seu peso? "))
altura = float(input("Qual a sua altura? "))
imc = massa / (altura * altura)
if(imc < 18.5):
  print(f"Você está abaixo do peso, seu imc é {round(imc,2)}")
elif(imc >= 18.5 and imc < 24.9):
  print(f"Você está no peso normal, seu imc é {round(imc,2)}")
elif(imc >= 25 and imc <= 29.9):
  print(f"Você está obeso, seu imc é {round(imc,2)}")
```
