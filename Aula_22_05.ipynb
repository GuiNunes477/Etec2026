```python
soma_notas = 0
contador = 1

while contador <= 4:
    nota = float(input(f"Digite a {contador}ª nota: "))
    soma_notas += nota
    contador += 1
media = soma_notas / 4
print(f"\nMédia do aluno: {media:.2f}")
if media >= 7.0:
    print("Você foi aprovado")
elif media >= 5.0:
    print("Você ficou de recuperação")
elif media < 0 or media > 10:
  print("Nota invalida")
else:
    print("Você foi reprovado")
```
