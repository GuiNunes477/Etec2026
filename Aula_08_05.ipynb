```python
# 1. conte até o número escolhido
num = int(input("Insira um número: "))
for i in range(-1, num):
  print(i + 1)
```

```python
# 2. Contagem regresiva
numero = int(input("Insira um número: "))
for i in range(numero + 1, 0, -1):
  print(i - 1)
```

```python
# 3. Numeros pares
numeral = int(input("Coloque um número: "))
for i in range(numeral + 1):
  if i % 2 == 0:
   print(i)
```

```python
# 4. Numeros Impares
numeral = int(input("Coloque um número: "))
for i in range(numeral + 1):
  if i % 2 != 0:
   print(i)
```
