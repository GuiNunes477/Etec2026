```python
# Exercício 1 — Contagem
num = int(input("Escolha um número: "))
contador = 1
while contador <= num:
  print(contador)
  contador = contador + 1
```

```python
# Exercício 2 — Contagem regressiva
num = int(input("Escolha um número: "))
contador = 1
while num >= contador:
  print(num)
  num = num - 1
```

```python
# Exercício 4 — Soma dos números
num = int(input("Escolha um número: "))
contador = 1
soma = 0
while contador <= num:
  soma += contador
  contador += 1
print(soma)
```

```python
# Exercício 5 — Tabuada
num = int(input("Escolha um número: "))
contador = 1
while contador <= 10:
  print(f"{num} x {contador} = {num * contador}")
  contador = contador + 1
```

```python
#Exercício 6 — Senha
senha = 1234
se = ''
while se != senha:
  se = int(input("Digite a senha: "))
  if se == senha:
   print("Senha correta")
  else:
   print("Senha incorreta")
```

```python
# Exercício 7 — Somando até digitar zero
soma = 0
numero = ''
while numero != 0:
  numero = int(input("Digite um número: "))
  soma = (numero + soma)
print(soma)
```

```python
# Exercício 8 — Maior número
numero = ''
maior = 0
while numero != 0:
  numero = int(input("Digite um número: "))
  if (numero > maior):
   maior = numero
print(maior)
```

```python
# Exercício 9 — Média das notas
contador = 0
nota = ''
media = 0
notas = 0
while contador <= 4:
  nota = int(input("Qual foi a sua nota? "))
  notas = (nota + notas)
  media = (notas / 5)
  contador = contador + 1
print(f"Sua média é {media}")
```
