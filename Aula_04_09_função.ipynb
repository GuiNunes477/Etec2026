```python
# 9. Positivo ou negativo
numero = float(input("Insira um número "))
def verificar_numero(numero):
  if numero > 0:
    print(f"o número {numero} é positivo")
  elif numero < 0:
    print(f"O número {numero} é negativo")
  else:
    print(f"Seu número é zero")
verificar_numero(numero)
```

```python
# 10. Média
nota1 = float(input("Qual foi a sua primeira nota? "))
nota2 = float(input("Qual foi a sua segunda nota? "))
def calcular_media(nota1,nota2):
  print(f"Sua média foi de {round((nota1 + nota2)/2,2)}")
calcular_media(nota1, nota2)
```
